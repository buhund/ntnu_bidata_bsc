```python
# Q01.01 - Beregn foroverdifferansen av f(x) i punkt x for skrittlengde h
# Foroverdifferanse er en metode for å beregne den deriverte av en funksjon.
# Gir en tilnærming av den deriverte.
# Vi tar differansen mellom funksjonsverdien i x, og funksjonsverdien i et punkt litt lenger bort, x+h.
# (f(x + h) - f(x)) / h

def f(x):
    return np.sin(x)

x = -5
h= 0.1 # 1/10

foroverdifferanse = (f(x + h) - f(x)) / h


print(f"{foroverdifferanse:.4f}")
```

```python
# Q01.02 - Beregn bakoverdifferansen av f(x) i punkt x for skrittlengde h
# (f(x) - f(x - h)) / h

def f(x):
    return x**3 - x + 1

x = -3
h = 1/5

bakoverdifferanse = (f(x) - f(x - h)) / h

print(f"{bakoverdifferanse:.4f}")
```

```python
# Q01.03 - Beregn senterdifferansen av f(x) i punkt x for skrittlengde h
# (f(x + h) - f(x - h)) / (2*h)

def f(x):
    return np.log(1+x)

x = 1
h = 3/10

senterdifferanse = (f(x + h) - f(x - h)) / (2*h)

print(f"{senterdifferanse:.4f}")
```

Her skal du få kode din egen økonomikommentator som kommenterer prisveksten fra måned til måned (selv om man i avisene gjerne tar endringen av KPI fra samme måned i fjor til denne måneden, og ser på endringen av denne fra måned til måned).

Skriv ferdig funksjonen kommenter(p) som tar inn en liste på samme form som KPI, går gjennom denne og skriver ut "Prisveksten avtok i april 2022." dersom differansen ovenfor er negativ for denne måneden, "Prisveksten økte i april 2022." dersom differansen er positiv, og "Prisveksten var uendret i april 2022." dersom differansen er null.

I tillegg, dersom differansen har skiftet fortegn skal den skrive ut en ekstra linje: "Nå snur det, hurra!" dersom den skifter fra positiv til negativ, og "Nå snur det, akk og ve!" dersom den skifter fra negativ til positiv.
For enkelhets skyld skal den altså ikke skrive ut noe om differansen skifter fortegn over flere enn to punkter (måneder) ved at den er null i de mellomliggende punktene.

Merk: Siden differansen (1) bruker forrige og neste verdi får vi ikke vurdert intervallene mars 2022 til april 2022 (fordi vi ikke har verdien fra februar 2022), og tilsvarende heller ikke perioden februar 2024 til mars 2024 (siden vi ikke har verdien i april 2024).

```python
KPI = [["mars 2022", 119.8],["april 2022", 121.2],["mai 2022", 121.5],["juni 2022", 122.6],["juli 2022", 124.2], ["august 2022", 123.9], ["september 2022", 125.6], ["oktober 2022", 126.0],
       ["november 2022",125.8],["desember 2022", 125.9],["januar 2023", 126.1],["februar 2023", 126.6],["mars 2023", 127.6],["april 2023", 129.0],["mai 2023", 129.6],["juni 2023", 130.4],
       ["juli 2023", 130.9], ["august 2023", 129.9], ["september 2023", 129.8], ["oktober 2023", 131.1],["november 2023",131.8],["desember 2023", 131.9],["januar 2024", 132.0],
       ["februar 2024", 132.3],["mars 2024", 132.6]]


def kommenter(p):
    prev_diff = None  # Variabel som holder på forrige differanse.
    for i in range(1, len(p)-1):
        # Beregner den andrederiverte til KPI
        diff = (p[i+1][1] - 2 * p[i][1] + p[i-1][1])

        # Print kommentar basert på diff
        if diff < 0:
            print(f"Prisveksten avtok i {p[i][0]}.")
        elif diff > 0:
            print(f"Prisveksten økte i {p[i][0]}.")
        else:
            print(f"Prisveksten var uendret i {p[i][0]}.")

        # Sjekker om diff har skiftet fortegn fra forrige måned.
        if prev_diff is not None:
            if prev_diff > 0 and diff < 0:
                print("Nå snur det, hurra!")
            elif prev_diff < 0 and diff > 0:
                print("Nå snur det, akk og ve!")

        # Oppdaterer prev_diff for neste runde.
        prev_diff = diff

# Kall funksjonen for å se resultatet
kommenter(KPI)
```

```python
# Q8

import numpy as np

def trapes(f, a, b, n):
    # Beregn skrittlengden (delintervall-bredde)
    h = (b - a) / n

    # Evaluer funksjonen i start- og sluttpunktet
    integral = 0.5 * (f(a) + f(b))

    # Legg til summen av funksjonsverdiene på de indre punktene
    for i in range(1, n):
        x_i = a + i * h
        integral += f(x_i)

    # Multipliser med skrittlengden h
    integral *= h

    return integral

# Definer funksjonen f(x) = sin(x^2)
def f(x):
    return np.sin(x**2)

# Bruk trapesmetoden til å tilnærme integralet av f(x) over [0, 1] med n = 10
T = trapes(f, 0, 1, 10)
print(f"Tilnærmet verdi av integralet: {T}")
```
