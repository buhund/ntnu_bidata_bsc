## Oppgave 1

I denne øvingen skal du finne fikspunktene til ulike funksjoner g=g(x). Her kan det være mer enn én løsning, så svaret skal skrives inn som en mengde: det vil si, dersom både 2 og 3 er fikspunkter for samme funksjon skal du oppgi svaret som {2,3}. Rekkefølgen har ingenting å si.

```python
import numpy as np # importer matematiske funksjoner etc.

def g1(x):
    return 6/x

def g2(x):
    return x**17

def g3(x):
    return (8 + 5*x) / (5 + x**2)

def fixpoint(g,x0,tol,maxIt):
    # implementasjon av fikspunktalgoritmen for en funksjon g,
    # startgjett x0, feiltoleranse tol and maksimalt antall iterasjoner maxIt.
    # Returnerer siste iterat, feilen and antall iterasjoner
    x_prev = x0
    x_next = g(x_prev)
    It = 1
    err = np.abs(x_next-x_prev)
    while err > tol and It < maxIt:
        x_prev = x_next
        x_next = g(x_next)
        err = np.abs(x_next-x_prev)
        It += 1
    return x_next, err, It

# La oss sjekke at vi får det samme svaret som i figurene i eksempel 2 ved å velge x0 = 1.5 og maxIt = 2
x0 = 1.5
iterations = 2
print('Med startpunkt {} og to iterasjoner med g1 får vi {}'.format(x0,fixpoint(g1,x0,1e-3,iterations)[0]))
print('Med startpunkt {} og to iterasjoner med g2 får vi {}'.format(x0,fixpoint(g2,x0,1e-3,iterations)[0]))
```

```python
# Del 2
import math

# Definere funksjonen og dens deriverte
def f(x):
    return math.exp(x) - (3 - x)

def df(x):
    return math.exp(x) + 1

# Newtons metode
def newton_method(x0, iterations):
    x = x0
    for _ in range(iterations):
        x = x - f(x) / df(x)
    return x

# Startverdi og antall iterasjoner (x2 er etter 2 iterasjoner)
x0 = 0
iterations = 2
x2 = newton_method(x0, iterations)

# Svar med 5 gjeldende desimaler

print(f"Verdien av x_2 = {round(x2, 5)}")
```

```python
# Del 2

# Vi kjører Newtons metode med flere iterasjoner for å få en stabil verdi av a
iterations = 10  # Øk antall iterasjoner for å stabilisere verdien
a = newton_method(x0, iterations)

# Svar med 5 gjeldende desimaler
print(f"Verdien av a = {round(a, 5)}")
```

Her vil vi tilnærme løsningen av ligningen
$$
2x=cos(x)

med fikspunktiterasjon, og velger oss fikspunkt-funksjonen 
$$
g(x)=cos(x)/2
$$

Her velger vi startgjetningen $x_0 = 0$ som gir et avvik $|2 \cdot 0 − cos(0)| = 1$

Vi ønsker å iterere frem til avviket for iterat $x_n$ er mindre enn eller lik $10^{−5}$, bruk feltet nedenfor til å kode dette.

Her er det viktig at du lagrer det siste iteratet som variabelen `x` og antall iterasjoner som variabelen `n`.

Hint: Her kan du få bruk for det du kan om løkker fra programmering.

```python
import numpy as np

# Definer funksjonen g(x) = cos(x) / 2
def g(x):
    return np.cos(x) / 2

# Startgjetning
x = 0

# Maksimumsavvik
tolerance = 10**(-5)

# Start antall iterasjoner
n = 0

# Iterer til avviket er mindre enn toleransen
while abs(2 * x - np.cos(x)) > tolerance:
    x = g(x)
    n += 1

# Siste iterat og antall iterasjoner er lagret i x og n
x, n
```

## Spørsmål 7
Et generelt tredjegradspolynom kan skrives på formen
$$
p(x)=c_0 + c_1x + c_2x^2 + c_3 x^3
$$

for koeffisienter $c_i , i = 0, 1, 2, 3$

I $x = 1$ tar dette polynomet verdien $p(1)$, slik at grafen til polynomet går gjennom punktet $(x, y) = (1, p(1))$

Verdien $p(1)$  er gitt av
$$
p(1) = c_0 + c_1 + c_2 + c_3
$$

som kan skrives som vektorproduktet

$$
p(1) = [1 1 1 1] 
\begin{bmatrix}
c_0 \\
c_1 \\
c_2 \\
c_3
\end{bmatrix} = [1 1 1 1]c
$$


Nå vil vi finne koeffisientene $c_i$ som gjør at polynomet går gjennom de fire punktene

$(−1, −2)$, $(0, 1)$, $(1, −1)$ og $(2, 4)$


Hvert punkt gir oss en ligning, og disse fire ligningene kan skrives som et lineært system $Ac = p$

for en $4 \times 4$-matrise $A$ og en vektor $p$ av polynomverdier.

Nedenfor skal du sette opp matrisen `A` og vektoren `p`, hvor vektoren skal være på formen

$$
p = 
\begin{bmatrix}
p(-1) \\
cp(0) \\
p(1) \\
p(2)
\end{bmatrix}
$$

og bruk deretter `np.linalg.solve` til å løse dette systemet og lagre svaret som variabelen `c`. 

```python
import numpy as np

# Set opp matrisen A basert på punktene (-1, -2), (0, 1), (1, -1), (2, 4)
A = np.array([
    [1, -1, (-1)**2, (-1)**3],  # For x = -1
    [1, 0, 0**2, 0**3],         # For x = 0
    [1, 1, 1**2, 1**3],         # For x = 1
    [1, 2, 2**2, 2**3]          # For x = 2
])

# Set opp vektoren p som inneholder verdiene av polynomet på punktene
p = np.array([-2, 1, -1, 4])

# Løs systemet Ac = p ved bruk av np.linalg.solve
c = np.linalg.solve(A, p)

# Vis løsningen
print(f"Koeffisientene til polynomet er: {c}")
```

## Spørsmål 8

For skalare ligninger $f(x) = 0$ med en skalar ukjent $x$ har vi sett at fikspunktmetoden tar utgangspunkt i å skrive om ligningen på formen $x = g(x)$.

Deretter definerer vi en fikspunkt-iterasjon gjennom $x_{n+1} = g(x_n)$.

Vi kan forsøke å bruke samme idé på det lineære systemet $Ax = b$, hvor den ukjente nå er en vektor $x$.

Dette kan vi gjøre ved å skrive $A$ som $A = D + (A - D)$, hvor $D$ er matrisen med samme diagonalelementer som $A$ og resten av elementene er null. Da får vi

$$
b = Ax = Dx + (A - D)x.
$$

Med dette uttrykket kan vi lage oss en fikspunktiterasjon

$$
D x_{n+1} = b - (A - D)x_n.
$$

Så lenge diagonalelementene i $D$ ikke er null kan vi løse dette systemet for $x_{n+1}$.

Gitt

$$
A =
\begin{bmatrix}
3 & 4 & 5 \\
1 & 6 & 2 \\
2 & 0 & 9
\end{bmatrix}, \quad b =
\begin{bmatrix}
1 \\
2 \\
3
\end{bmatrix},
$$

fullfør koden under slik at funksjonen `g(x)` tar inn en 3-vektor (et array) $x$ og returnerer løsningen av ligningen (1) for $A$ og $b$ ovenfor.

Hint: Her kan funksjonen `np.linalg.solve` være til hjelp.

Dette er såkalt Jacobi-iterasjon, en fikspunktmetode for å løse lineære systemer.

```python
import numpy as np

# Definer matrisen A og vektoren b
A = np.array([[3, 1, 2], [4, 6, 0], [5, 2, 9]])  # Riktig matrise
b = np.array([1, 2, 3])

# Trekk ut diagonalmatrisen D fra A
D = np.diag(np.diag(A))

# Definer funksjonen g(x) for Jacobi-iterasjon
def g(x):
    # Beregn høyresiden av ligningen b - (A - D)x
    rhs = b - np.dot(A - D, x)
    # Løs Dx_{n+1} = rhs for x_{n+1}
    return np.linalg.solve(D, rhs)

# Startgjetning (nullvektor)
x0 = np.zeros(3)

# Utfør Jacobi-iterasjon for noen få iterasjoner
x = x0
iterations = 5
for _ in range(iterations):
    x = g(x)

# Vis siste verdi av x etter 5 iterasjoner
x
```
