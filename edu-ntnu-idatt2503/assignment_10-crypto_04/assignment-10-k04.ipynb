```python
# 1b

def extended_gcd(a, b):
    """
    Extended Euclidean Algorithm
    Returns a tuple of (gcd, x, y), where gcd is the greatest common divisor of a and b,
    and x, y are the coefficients satisfying Bézout's identity: ax + by = gcd(a, b)
    """
    if a == 0:
        return (b, 0, 1)
    else:
        gcd, x1, y1 = extended_gcd(b % a, a)
        # Update x and y using the results of recursion
        x = y1 - (b // a) * x1
        y = x1
        return (gcd, x, y)

def modinv(a, m):
    """
    Computes the modular inverse of a modulo m
    """
    gcd, x, y = extended_gcd(a, m)
    if gcd != 1:
        raise Exception(f"Modular inverse does not exist for a = {a} and modulus m = {m}")
    else:
        return x % m

# Given values
p = 1283
q = 2027
d = 3

# Compute n and phi(n)
n = p * q
phi_n = (p - 1) * (q - 1)

# Compute e such that e * d ≡ 1 mod phi(n)
try:
    e = modinv(d, phi_n)
    print(f"Public exponent e is: {e}")
except Exception as ex:
    print(ex)

# Verify that (e * d) % phi_n == 1
verification = (e * d) % phi_n
print(f"Verification (e * d) mod phi(n) = {verification}")
```

```python
# 1c

def modular_exponentiation(base, exponent, modulus):
    """
    Performs modular exponentiation using the repeated squaring method.
    Computes (base ** exponent) % modulus efficiently.
    """
    result = 1
    base = base % modulus  # Ensure base is within modulus

    while exponent > 0:
        # If exponent is odd, multiply the result by the current base
        if exponent % 2 == 1:
            result = (result * base) % modulus
        # Square the base
        base = (base * base) % modulus
        # Halve the exponent (integer division)
        exponent = exponent // 2

    return result

# Given values
m = 111  # Message to encrypt
e = 1731555  # Public exponent
n = 2600641  # Modulus

# Encrypt the message
ciphertext = modular_exponentiation(m, e, n)
print(f"The encrypted message (ciphertext) is: {ciphertext}")


# Decryption using the private key
d = 3  # Private exponent
decrypted_message = modular_exponentiation(ciphertext, d, n)
print(f"The decrypted message is: {decrypted_message}")
```
